```python
import torch
from torch import nn

x = torch.tensor([1., 3., 5., 7.]).reshape(2, 1, 1, 2)
bn = nn.BatchNorm2d(1)

print("Kształt:", x.shape)
print("Średnia kanału:", x.mean(dim=(0, 2, 3)))
print("Wariancja kanału:", x.var(dim=(0, 2, 3), unbiased=False))

bn.train()
y_train = bn(x)

print("Wynik train:", y_train.detach())
print("Zapamiętana średnia:", bn.running_mean)
print("Zapamiętana wariancja:", bn.running_var)

bn.eval()
with torch.no_grad():
    y_eval = bn(x)

print("Wynik eval:", y_eval)
```
